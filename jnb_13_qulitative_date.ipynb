```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
xls = pd.ExcelFile('Superstore.xlsx')
sheet_names = xls.sheet_names
sheet_names
```

```python
sales = {}
for name in sheet_names:
    sales[name] = pd.read_excel('Superstore.xlsx', sheet_name=name)
```

### 1. Determine whether there is an association between Segment and Region. If so, how strong is the association?


```python
segment = sales['Orders']['Segment']
region = sales['Orders']['Region']
```

```python
orders = sales['Orders']
```

```python
orders.columns
```

```python
# Contingency table: Analyze the frequency of each Segment–Region combination

contingency_table = (
    orders
    .groupby(['Segment', 'Region'])
    ['Order ID']
    .nunique()
    .reset_index()
    .pivot(index='Segment', columns='Region')
)
contingency_table
```

```python
pd.crosstab(orders['Segment'], orders['Region'])
```

```python
stats.chi2_contingency(contingency_table.values)
```
