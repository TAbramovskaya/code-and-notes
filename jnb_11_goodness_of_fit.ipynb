```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import scikit_posthocs as sp
```

```python
# PSES p-402 ex-10.79

obs = np.array([98, 162, 180, 60])
ratio = np.array([3, 5, 6, 2])
exp = ratio * sum(obs) / sum(ratio)
```

```python
stats.chisquare(obs, exp)
```

```python
# PSES p-402 ex-10.81

obs = [32, 38, 46, 44, 34, 46]
exp = np.ones(6) * sum(obs) / 6
```

```python
stats.chisquare(obs, exp)
```

```python
# PSES p-402 ex-10.82
vals = [0, 1, 2, 3]
obs = [1, 31, 55, 25]

h_distr = stats.hypergeom(8, 5, 3)

trials = sum(obs)
exp = [h_distr.pmf(v) * trials for v in vals]
```

```python
exp
```

```python
obs = [obs[0] + obs[1]] + obs[2:]
exp = [exp[0] + exp[1]] + exp[2:]
```

```python
stats.chisquare(obs, exp)
```
