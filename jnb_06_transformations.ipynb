```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Salary.csv')
```

```python
data.info()
```

```python
salary = data['Зарплата'].copy()
```

```python
transformations = [lambda x: x, np.log, np.log10, np.sqrt, np.cbrt, lambda x: 1/x]
names = ['original', 'log', 'log10', 'square root', 'cube root', 'inverse']
salary_transformed = dict()

for name, transformation in zip(names, transformations):
    salary_transformed[name] = transformation(salary)
```

```python
fig, axes = plt.subplots(1, len(transformations), figsize=(17, 5))

for ax, name in zip(axes, names):
    ax.hist(salary_transformed[name])
    ax.set_title(name)
    ax.set_xlabel('Salary')
    ax.set_ylabel('Frequency')
plt.show()
```
