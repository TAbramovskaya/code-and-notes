```python
# PSCS e-342, p-362

pH_meter = np.array([7.07, 7, 7.1, 6.97, 7, 7.03, 7.01, 6.98, 7.08])

n = len(pH_meter)
mu_0 = 7
```

```python
pH_meter.mean()
```

```python
describe_result = stats.describe(pH_meter)

for field, value in zip(describe_result._fields, describe_result):
    print(f'{field}: {value}')
```

```python
ttest_result = stats.ttest_1samp(a=pH_meter, popmean=mu_0)

for field, value in zip(ttest_result._fields, ttest_result):
    print(f"{field}: {value}")
```

```python
alpha = 0.05
t_distribution = stats.t(df=n-1)
print(f"Critical values for the {1 - alpha} significance level: {t_distribution.ppf(alpha/2), t_distribution.ppf(1 - alpha/2)}")
```

```python
ttest_result.confidence_interval(confidence_level=0.95)
```

```python
se = stats.sem(pH_meter)

ci = stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=describe_result.mean,
    scale=se
)

print(ci)
```

```python
before = np.array([2.76, 5.18, 2.68, 3.05, 4.10, 7.05, 6.60, 4.79, 7.39, 7.30, 11.78, 3.90, 26.00, 67.48, 17.04])

after = np.array([7.02, 3.10, 5.44, 3.99, 5.21, 10.26, 13.91, 18.53, 7.91, 4.85, 11.10, 3.74, 94.03, 94.03, 41.7])

n = len(before)
diff = after - before
```

```python
alpha = 0.05

lower = alpha / 2
upper = 1 - alpha / 2

t_distribution = stats.t(df=n-1)
critical_values = (t_distribution.ppf(lower), t_distribution.ppf(upper))
critical_values
```

```python
injection_diff = stats.describe(diff)

for field, value in zip(injection_diff._fields, injection_diff):
    print(f"{field}: {value}")

print("sample standard deviation:")
print(diff.std(ddof=1))
print(math.sqrt(injection_diff.variance))
```

```python
injection_test_result = stats.ttest_1samp(diff, popmean=0)
for field, value in zip(injection_test_result._fields, injection_test_result):
    print(f"{field}: {value}")
```

```python
# PSES p-375, e-355 (ex 9.40 p-314, e-294)
no_nitrogen = np.array([0.53, 0.32, 0.28, 0.37, 0.47, 0.43, 0.36, 0.42, 0.38, 0.43])
nitrogen = np.array([0.26, 0.43, 0.47, 0.49, 0.52, 0.75, 0.79, 0.86, 0.62, 0.46])

n = len(no_nitrogen)
```

```python
nitrogen_test_result_equal = stats.ttest_ind(nitrogen, no_nitrogen, equal_var=True, alternative="greater")

for field, value in zip(nitrogen_test_result_equal._fields, nitrogen_test_result_equal):
    print(f"{field}: {value}")
print(f"DoF: {nitrogen_test_result_equal.df}")
```

```python
nitrogen_test_result_unequal = stats.ttest_ind(nitrogen, no_nitrogen, equal_var=False, alternative="greater")

for field, value in zip(nitrogen_test_result_unequal._fields, nitrogen_test_result_unequal):
    print(f"{field}: {value}")
print(f"DoF: {nitrogen_test_result_unequal.df}")
```

```python
def pprint(fields, values):
    for field, value in zip(fields, values):
        print(f"{field}: {value}")
    print(f"standard deviation: {math.sqrt(values.variance)}")
    print(f"standard error: {math.sqrt(values.variance) / math.sqrt(len(values))}")
```

```python
no_nitrogen_describe = stats.describe(no_nitrogen)

pprint(no_nitrogen_describe._fields, no_nitrogen_describe)
```

```python
nitrogen_describe = stats.describe(nitrogen)

pprint(nitrogen_describe._fields, nitrogen_describe)
```
